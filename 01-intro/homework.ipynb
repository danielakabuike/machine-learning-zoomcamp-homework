```python
df = pd.read_csv('https://raw.githubusercontent.com/DataTalksClub/machine-learning-zoomcamp/main/cohorts/2026/data/car_fuel_efficiency_2026.csv')
```

```python
# Returns an array of the distinct fuel types
distinct_fuels = df['fuel_type'].unique()
print(distinct_fuels)
```

```python
# Returns the total number of columns with at least one missing value
num_cols_with_nulls = df.isna().any().sum()
print("Columns with missing values:", num_cols_with_nulls)
```

```python
# Filter for cars originating from Asia and find the maximum mpg
max_asia_mpg = df[df['origin'] == 'Asia']['fuel_efficiency_mpg'].max()

print("Maximum fuel efficiency of cars from Asia:", max_asia_mpg)
```

```python
# 1. Calculate the initial median
initial_median = df['horsepower'].median()
print(f"Initial Median: {initial_median}")

# 2. Find the most frequent value (mode)
# .mode()[0] extracts the single most frequent value
mode_val = df['horsepower'].mode()[0]
print(f"Most Frequent Value (Mode): {mode_val}")

# 3. Fill missing values with the mode
# (Using a copy or temporary series so you don't accidentally overwrite if testing)
horsepower_filled = df['horsepower'].fillna(mode_val)

# 4. Calculate the new median
new_median = horsepower_filled.median()
print(f"New Median: {new_median}")

# 5. Check if it changed
if new_median > initial_median:
    print("Verdict: Yes, it increased")
elif new_median < initial_median:
    print("Verdict: Yes, it decreased")
else:
    print("Verdict: No")
```

```python
import numpy as np

# 1. Select all the cars from Asia
df_asia = df[df['origin'] == 'Asia']

# 2. Select only columns 'vehicle_weight' and 'model_year'
df_selected = df_asia[['vehicle_weight', 'model_year']]

# 3. Select the first 7 values
df_first7 = df_selected.head(7)

# 4. Get the underlying NumPy array X
X = df_first7.to_numpy() # or df_first7.values

# 5. Compute matrix multiplication X^T * X
XTX = X.T @ X

# 6. Invert XTX
XTX_inv = np.linalg.inv(XTX)

# 7. Create array y
y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])

# 8. Multiply the inverse of XTX with X^T, then by y to get weights w
w = (XTX_inv @ X.T) @ y

# 9. Compute the sum of all elements of w
sum_of_weights = w.sum()

print("Weights (w):", w)
print("Sum of weights:", sum_of_weights)
```
